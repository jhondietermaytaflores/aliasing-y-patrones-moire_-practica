```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img1 = cv2.imread("1.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("2.jpg", cv2.IMREAD_GRAYSCALE)
img3 = cv2.imread("3.jpg", cv2.IMREAD_GRAYSCALE)

def aplicar_gaussian_blur(img):
    return cv2.GaussianBlur(img, (5, 5), 0)

blur1 = aplicar_gaussian_blur(img1)
blur2 = aplicar_gaussian_blur(img2)
blur3 = aplicar_gaussian_blur(img3)


kernel_diagonal = np.array([
    [-1,  0,  1],
    [ 0,  0.5,  0],
    [ 1,  0, -1]
])

def aplicar_convolucion(img, kernel):
    conv = cv2.filter2D(img, -1, kernel)
    conv_normalizada = cv2.normalize(conv, None, 0, 255, cv2.NORM_MINMAX)
    return conv_normalizada

conv1 = aplicar_convolucion(blur1, kernel_diagonal)
conv2 = aplicar_convolucion(blur2, kernel_diagonal)
conv3 = aplicar_convolucion(blur3, kernel_diagonal)


def mostrar_resultados(original, blur, conv, titulo):
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 3, 1)
    plt.title(f"{titulo} - Original")
    plt.imshow(original, cmap='gray')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.title(f"{titulo} - Gaussian blur")
    plt.imshow(blur, cmap='gray')
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    plt.title(f"{titulo} - Convolución Diagonal")
    plt.imshow(conv, cmap='gray')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()

mostrar_resultados(img1, blur1, conv1, "Imagen 1")
mostrar_resultados(img2, blur2, conv2, "Imagen 2")
mostrar_resultados(img3, blur3, conv3, "Imagen 3")
```
